```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
value=pd.read_csv("C:/Users/ELCOT/Downloads/database.csv (1).zip")
```

```python
print(value.head())
```

```python
print(value.isnull().sum())
```

```python
value=value.dropna()
```

```python
print(value.isnull().sum())
```

```python
print(value.duplicated())
```

```python
value.duplicated().sum()
```

```python
value.drop_duplicates()
```

```python
value["Date"]=pd.to_datetime(value["Date"])
```

```python
print(value.dtypes)
```

```python
plt.plot(value["Date"],value["Latitude"])
plt.show()
```

```python
value["Moving_Average"]=value["Latitude"].rolling(window=3).mean()
```

```python
print(value[["Date","Latitude","Moving_Average"]].head(10))
```

```python
plt.plot(value["Date"],value["Latitude"], label="Latitude")
plt.plot(value["Date"],value["Moving_Average"], label="Moving Average")
plt.xlabel("Date")
plt.ylabel("Latitude")
plt.title("Moving Average")
plt.legend()
plt.show()
```

```python
value1=value["Date"].iloc[0]
value2=value["Date"].iloc[-1]

if value2 > value1:
    print("Increasing Trend")
elif  value2 < value1:
    print("Decreasing Trend")
else:
    print("Stable Trend")
```
